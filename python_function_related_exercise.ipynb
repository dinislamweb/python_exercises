name = "Kelly"<br>

2. Create a function func1() such that it can accept a variable number of arguments and print all of them. Whether you pass two numbers or five, the function should handle them all without error.Given Input:

Call 1: func1(20, 40, 60)
Call 2: func1(80, 100)<br>
Expected Output:

Printing values:
20
40
60
Printing values:
80
100

```python
def func(*numbers):
    for i in numbers:
        print(i,end=" ")
    print()
func(20,40,40)
func(80,10)
```

3. Write a function calculation() that accepts two variables and calculates both addition and subtraction. The function must return both results in a single return statement.
<br>Given Input:

a = 40
b = 10<br>
Expected Output: 50, 30

```python
def calculation(num1, num2):
    add = num1+num2
    sub = num1-num2
    return add,sub
a = 40
b = 10
print(calculation(a,b))
```

4. Create a function show_employee() that accepts an employee’s name and salary. If the salary is not provided in the function call, the function should automatically assign a default value of 9000.<br>
Given Input:

Case 1: name="Ben", salary=12000<br>
Case 2: name="Jessa" (salary missing)<br>
Expected Output:<br>

Name: Ben salary: 12000<br>
Name: Jessa salary: 9000

```python
def show_employee(name,salary=9000):
    print(f"Name: {name} salary: {salary}")
show_employee("Ben",12000)
show_employee("Jessa")
```

5. Create an outer function that accepts two parameters, a and b. Inside, create an inner function that calculates the addition of a and b. The outer function should then add 5 to that sum and return the final result.<br>
a = 5<br>
b = 10<br>
Expected Output: 20

```python
def outer_func(a,b):
    def inner_func(a,b):
        return a+b
    add = inner_func(a,b)+5
    return add
   
print(outer_func(5,10))
```
