12. Write a program that counts the total number of vowels and consonants in a given sentence, ignoring spaces and special characters.

```python
sentence = "Loops are Fun!"
vowels = "aeiou"
v_count = 0
c_count = 0

for char in sentence.lower():
    if char.isalpha():  # Only process letters
        if char in vowels:
            v_count += 1
        else:
            c_count += 1

print(f"Vowels: {v_count}")
print(f"Consonants: {c_count}")
```

14. Write a program to find the largest and smallest digit within a given integer (e.g., in 75869, the largest is 9 and the smallest is 5).

```python
num = 75869
smallest = 9
largest = 0
while num>0:
    digit = num % 10
    if digit > largest:
        largest = digit
    if digit < smallest:
        smallest = digit
    num//=10
print(f"Smallest digit: {smallest}")
print(f"Largest digit:{largest}")
        
```

15. Collatz Conjecture: Generate a sequence until it reaches 1. The Collatz conjecture states that if you start with any positive integer n, and if n is even, divide it by 2; if n is odd, multiply it by 3 and add 1. Repeat the process. The sequence will always eventually reach 1. Write a program to print this sequence for a given number.

```python
n = 6

while n != 1:
    print(n,end=" ")
    if n % 2 == 0:
       n = n //2
    else:
        n = 3*n + 1

print(1)  
    
```

16. Write a program to check if a number is an Armstrong number. An Armstrong number (for a 3-digit number) is an integer such that the sum of the cubes of its digits is equal to the number itself (e.g., 153 = 1^3 + 5^3 + 3^3).

```python
num = int(input())
original = num
power = len(str(num))
sum = 0
while num>0:
    digit = num % 10
    sum = sum + digit **power
    num//=10
if original == sum:
    print(f"{original} is an Armstrong number")
else:
    print(f"{original} is not an Armstrong number")
```

20.  Print a 5*5 square of stars where the middle is empty, leaving only the border.

```python
n = 5
for i in range(n):
    for j in range(n):
        if i==0 or i == n-1 or j==0 or j==n-1:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```

21. Write a program to Print pyramid pattern of stars using nested loops:

```python
n = 5
for i in range(n):
    for j in range(i):
        print("*",end=" ")
    print()
for i in range(n):
    for j in range(n-i):
        print("*",end=" ")
    print()
```

22. Given a dictionary of student scores, create a new dictionary that only includes students who scored above a certain threshold (e.g., 75).
scores = {"Alice": 85, "Bob": 70, "Charlie": 95, "David": 60} threshold = 75
Expected Output:

Passing Students: {'Alice': 85, 'Charlie': 95}

```python
scores = {"Alice": 85, "Bob": 70, "Charlie": 95, "David": 60}
threshold = 75
frequency = {}
for name,score in scores.items():
    if score > threshold:
        frequency[name] = score
print(frequency)
```

25. Write a program to check if a number is a “Perfect Number.” A perfect number is a positive integer that is equal to the sum of its proper divisors (excluding the number itself). For example, 6 is perfect because 1 + 2 + 3 = 6.

```python
num = int(input())
original = num
sum = 0
for i in range(1,(num//2)+1):
    if num % i == 0:
        sum+=i
if original == sum:
    print(f"{original} is perfect number")
else:
    print(f"{original} is not a perfect number")
```

26. Write a program to display all prime numbers within a range (e.g., 25 to 50). A prime number is a natural number greater than 1 that is not a product of two smaller natural numbers.

```python
start = 50
end = 60
for i in range(start,end+1):
    if i == 1:
            continue
    for j in range(2,(i//2)+1):
        if i%j == 0:
            break
    else:
        print(i)
        
```

27. Write a program to calculate the sum of the series 2 + 22 + 222 + 2222 + …. up to N terms. For example, if n=5, the series is 2 + 22 + 222 + 2222 + 22222.

27. Given a 2D list (matrix), find the row and column index of a target value.<br>
matrix = [[10, 20], [30, 40], [50, 60]]<br>
target = 30<br>
Expected Output:<br>

Target 30 found at Row: 1, Column: 0

```python
matrix = [[10, 20], [30, 40], [50, 60]]
target = 30
for row_index,row in enumerate(matrix):
    for col_index, value in enumerate(row):
        if value == target:
            print(f"Row :{row_index}, Column:{col_index}")
```
