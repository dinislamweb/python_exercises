##  Remove Dictionary Items

```python
thisdict = {
    'brand' : 'Ford',
    'model' : 'Mostang',
    'year' : 1964
} 
thisdict.pop('model')
print(thisdict)
```

```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
thisdict.popitem()
print(thisdict)
```

### Nested Dictionaries
- A dictionary can contain dictionaries, this is called nested dictionaries.

```python
sisters = {
    "Ayesha":{
        "child1":"Mehedi",
        "child2" : "Sarowar",
        "child3" : "Tamim",
         "child4" : "Fatema"
    },
    "Tahmina" : {
        "child1" : "Hamida",
        "child2" : "Rabbi",
        "child3" : "Aliza"
    },
    "Monira":{
        "child1" : "Razim",
        "child2" : "Adiba",
        "child3" : "Rahim"
    }
}
print(sisters)
```

```python
sister1 = {
        "child1":"Mehedi",
        "child2" : "Sarowar",
        "child3" : "Tamim",
         "child4" : "Fatema"
    }
sister2 = {
        "child1" : "Hamida",
        "child2" : "Rabbi",
        "child3" : "Aliza"
}
sister3 = {
        "child1" : "Razim",
        "child2" : "Adiba",
        "child3" : "Rahim"
    }
sisters = {
   'Ayesha':sister1,
    'Tahmina':sister2,
    'Monira':sister3
}
print(sisters)
```

```python
sisters['Ayesha']['child1']
```

## Loop Through Nested Dictionaries

```python
for x, obj in sisters.items():
  print(x)

  for y in obj:
    print(y + ':', obj[y])
```

1. Perform basic dictionary operations<br>
Given:

my_dict = {'name': 'Alice', 'age': 35, 'city': 'New York'}<br>
Perform following operations on given dictionary

Add New Key-Value Pair: Add a new key-value pair, 'profession': 'Doctor', to the dictionary and print the updated dictionary.<br>
Modify Value: Change the value of the age key to 40 in the dictionary and print the updated dictionary.<br>
Access Key: Print the value associated with the city key.<br>
Expected Output:

Original dictionary: {'name': 'Alice', 'age': 35, 'city': 'New York'}

Updated dictionary after adding 'profession': {'name': 'Alice', 'age': 35, 'city': 'New York', 'profession': 'Doctor'}

Updated dictionary after modification: {'name': 'Alice', 'age': 40, 'city': 'New York', 'profession': 'Doctor'}

City: New York

```python
my_dict = {'name': 'Alice', 'age': 35, 'city': 'New York'}
print("Original dictionary: ",my_dict)
my_dict.update({'profession':'Doctor'})
print("Updated dictionary after adding 'profession:",my_dict)
my_dict['age'] = 40
print("Updated dictionary after age modification:",my_dict)
x = my_dict['city']
print("City:",x)
```

2. Perform dictionary operations <br>
Given:

my_dict = {'name': 'Alice', 'age': 35, 'city': 'New York', 'profession': 'Doctor'}<br>
Perform following operations on given dictionary

Remove Key-Value Pair : Remove the profession key-value pair from the dictionary.<br>
Get Items (Key-Value Pairs): Print all key-value pairs (items) in the dictionary.<br>
Check if Key Exists in the dictionary<br>
Expected Output:

Original dictionary: {'name': 'Alice', 'age': 35, 'city': 'New York', 'profession': 'Doctor'}

Updated dictionary after removing 'profession': {'name': 'Alice', 'age': 35, 'city': 'New York'}

Printing all key-value pairs:
name: Alice<br>
age: 35<br>
city: New York<br>

Does 'age' exist? True

```python
my_dict = {'name': 'Alice', 'age': 35, 'city': 'New York', 'profession': 'Doctor'}
print("Original dictionary:",my_dict)
my_dict.pop('profession')
print("Updated dictionary after removing 'profession':",my_dict)

print("Printing all key-value pairs:")
for key, value in my_dict.items():
  print(f"{key}: {value}")

def check_key_exists(dictionary, key_to_check):
  return key_to_check in dictionary

key1 = 'age'
print(f"Does '{key1}' exist? {check_key_exists(my_dict, key1)}")
```

3. Dictionary from Lists<br>
Write a Python program to convert two Python lists into a dictionary where elements from the first list become keys and elements from the second list become values.

keys = ['Ten', 'Twenty', 'Thirty']<br>
values = [10, 20, 30]<br>
Expected output:

{'Ten': 10, 'Twenty': 20, 'Thirty': 30}

```python
keys = ['Ten', 'Twenty', 'Thirty']
values = [10, 20, 30]
result = dict(zip(keys,values))
print(result)
    
```

```python
keys = ['Ten', 'Twenty', 'Thirty']
values = [10, 20, 30]
result = dict()
for i in range(len(keys)):
    result.update({keys[i]:values[i]})
print(result)
```

6. Count Character Frequencies <br>
Given a string, create a dictionary where keys are characters and values are their frequencies in the string.

Given:

string1 = 'Jessa'<br>
Expected Output:

Frequencies for 'Jessa': {'J': 1, 'e': 1, 's': 2, 'a': 1}

```python
string1 = 'Jessa'
char_count = dict()
for i in string1:
    count = string1.count(i)
    char_count[i] = count
print(char_count)
    
```

10. Initialize dictionary with default values<br>
In Python, we can initialize the keys with the same values.

Given:

employees = ['Kelly', 'Emma']<br>
defaults = {"designation": 'Developer', "salary": 8000}<br>
Expected output:

{'Kelly': {'designation': 'Developer', 'salary': 8000}, 'Emma': {'designation': 'Developer', 'salary': 8000}}

```python
employees = ['Kelly', 'Emma']
defaults = {"designation": 'Developer', "salary": 8000}
res = dict.fromkeys(employees, defaults)
print(res)
```

11. Write a Python program to create a new dictionary by extracting the mentioned keys from the below dictionary.

Given dictionary:

sample_dict = {
    "name": "Kelly",
    "age": 25,
    "salary": 8000,
    "city": "New york"}

# Keys to extract
keys = ["name", "salary"]<br>
Expected output:

{'name': 'Kelly', 'salary': 8000}

```python
sample_dict = { "name": "Kelly", "age": 25, "salary": 8000, "city": "New york"}
keys = ["name", "salary"]
d = dict()
for i in keys:
    d.update({i: sample_dict[i]})
print(d)
```

13. While we know how to check for a key’s presence in a dictionary, it’s sometimes necessary to determine if a specific value exists.

Write a Python program to check if the value 200 is present in the provided dictionary.

Given:

sample_dict = {'a': 100, 'b': 200, 'c': 300}<br>
Expected output:

200 present in a dict

```python
sample_dict = {'a': 100, 'b': 200, 'c': 300}
if 200 in sample_dict.values():
    print("200 present in a dict")
else:
    print("Not present")
```

14. Rename key of a dictionary <br>
Write a program to rename a key city to a location in the following dictionary.

Given:

sample_dict = {
  "name": "Kelly",
  "age":25,
  "salary": 8000,
  "city": "New york"
}<br>
Expected output:

{'name': 'Kelly', 'age': 25, 'salary': 8000, 'location': 'New york'}

```python
sample_dict = { "name": "Kelly", "age":25, "salary": 8000, "city": "New york" }
sample_dict.pop("city")
sample_dict.update({"location":"New york"})
print(sample_dict)
```

```python
sample_dict = { "name": "Kelly", "age":25, "salary": 8000, "city": "New york" }
sample_dict['location'] = sample_dict.pop('city')
print(sample_dict)
```
