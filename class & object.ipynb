## Basic example of class and object

```python
class Student:
    name = "Hridoy"
    age = 20
student1 = Student()
print(f"Name:{student1.name}")
print(f"Age:{student1.age}")
```

## Better example

```python
class Student:
    def __init__(self, name,student_id, age):
        self.Name = name
        self.Student_id = student_id
        self.Age = age
student1 = Student("Hridoy",223,20)
print(student1.Name)
print(student1.Student_id)
print(student1.Age)        
```

## One Class to multiple Object

```python
class Department:
    def __init__(self,name, code, year):
        self.name = name
        self.code = code
        self.year = year
dept1 = Department("CSE",180,2002)
dept2 = Department("ME",181,2003)
dept3 = Department("CE",182,2003)
print(dept1.name,dept1.code,dept1.year)
print(dept2.name,dept2.code,dept2.year)
print(dept3.name,dept3.code,dept3.year)
```

## Function Inside the class

```python
class Student:
    def __init__(self,name,id, age):
        self.name = name
        self.id = id
        self.age = age
    def show_info(self):
        print(self.name)
        print(self.id)
        print(self.age)
        print()
      
    
student1 = Student("Hridoy",223,25)
student2 = Student("Hridy",222,22)
student1.show_info()
student2.show_info()
```

## Real-Life Example: Car 

```python
class Car:
    def __init__(self,brand, model, color):
        self.brand = brand
        self.model = model
        self.color = color
    def start(self):
        print(f"The {self.brand} model {self.model} color {self.color} is starting.")
        print()
    def show(self):
        print(f"Brand : {self.brand}")
        print(f"Model : {self.model}")
        print(f"Color : {self.color}")

car1 = Car("Toyota", "Corolla", "White")
car2 = Car("Honda", "Civic", "Black")

car1.show()
car1.start()
car2.show()
car2.start()
```

## DSA Class/Object

```python
class Node:

    def __init__(self, data):
        self.data = data
        self.next = None


node1 = Node(10)
node2 = Node(20)

node1.next = node2

print(node1.data)
print(node1.next.data)
```

## Rectangle Calculator

```python
class Rectangle:
    def __init__(self,length, width):
        self.length = length
        self.width = width
    def area(self):
        x = self.length * self.width
        print(f"Area of rectangle is {x} unit square")
    def perimeter(self):
        peri = 2 *(self.length + self.width)
        print(f"Perimeter of rectangle is {peri} unit")
r1 = Rectangle(10,5)
r1.area()
r1.perimeter()
```

## Bank Account

```python
class BankAccount:
    def __init__(self,account_name, balance):
        self.account_name = account_name
        self.balance = balance
    def deposit (self, amount):
        self.amount = amount
    def show_balance(self):
        self.balance +=self.amount
        print(f"Current Balance: {self.balance}")
account1 = BankAccount("Hridoy",1000)
account1.deposit(500)
account1.show_balance()
```

## Employee Salary

```python
class Employee():
    def __init__(self,name,employee_id,salary):
        self.name = name
        self.employee_id = employee_id
        self.salary = salary
    def increase_salary(self, amount):
        self.salary+= amount
        
    def show_info(self):
        print(f"Name: {self.name}")
        print(f"Employee ID: {self.employee_id}")
        print(f"Salary: {self.salary}")

    

emp1 = Employee("Rahim",101,3000)
emp1.increase_salary(5000)
emp1.show_info()
        
```
