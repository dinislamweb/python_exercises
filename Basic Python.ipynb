Write a program to check if a given number is a palindrome (reads the same forwards and backwards).<br>
Given Input:

Case 1: number = 121<br>
Case 2: number = 125

```python
def check_pal(number):
    original = number
    rev = 0
    while number > 0 :
        digit = number % 10
        rev = rev*10 + digit
        number//=10
    if original == rev:
        return "Palindrome"
    else:
        return "Not Palindrome"
num = int(input())
print(check_pal(num))
```
