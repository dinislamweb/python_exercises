- Write a Python function that accepts two integer numbers. If the product of the two numbers is less than or equal to 1000, return their product; otherwise, return their sum.

```python
def prodsum(num1,num2):
    if num1 * num2 <= 1000:
        return num1 * num2
    else :
        return num1 + num2

x,y = map(int,input().split())
print(prodsum(x,y))
x,y = map(int,input().split())
print(prodsum(x,y))
```

- Iterate through the first 10 numbers (0–9). In each iteration, print the current number, the previous number, and their sum.

```python
previous_num = 0
for i in range(10):
    current_num = i
    print(f"current number {current_num} previous number {previous_num} sum : {current_num+previous_num}")
    previous_num = current_num
```

Write a program that calculates the factorial of a given number (e.g., 5!) using a for loop.

```python
number = int(input())
fact = 1
for i in range(1,number+1):
    fact = fact*i
print(fact)
```

Write a program to count the total number of vowels (a, e, i, o, u) present in a given sentence.<br>
Given Input: sentence = "Learning Python is fun!"

Expected Output: Number of vowels: 6

```python
vowels = "aeiou"
sentence = "Learning Python is fun!"
count = 0
for i in sentence:
    if i in vowels:
        count+=1
print(f"Number of vowels: {count}")
```

Given a list of integers, find and print both the largest and the smallest numbers.<br>
nums = [45, 2, 89, 12, 7]

Expected Output: Largest: 89 Smallest: 2

```python
nums = [45, 2, 89, 12, 7]

largest = nums[0]
smallest = nums[0]

for num in nums:
    if num > largest:
        largest = num
    if num < smallest:
        smallest = num

print("Largest:", largest)
print("Smallest:", smallest)
```
